# YOLO 器械识别冒烟测试

使用 `E:\\miniconda\\envs\\yolo` 环境运行。这个 Notebook 只用于验证 `best.pt` 能否加载和完成一次预测，后续真实接口会按同样的输入输出接入后端。

```python
from pathlib import Path
from ultralytics import YOLO

model_path = Path(r"E:\gym_equipment\yolo11_runs\gym_yolo113\weights\best.pt")
assert model_path.exists(), model_path

model = YOLO(str(model_path))
print(model.names)
```

```python
sample_image = Path(r"E:\gym_equipment\images\拍照识别器械页.png")
results = model.predict(str(sample_image), conf=0.25)

for result in results:
    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        print({
            "class_id": class_id,
            "label": model.names[class_id],
            "confidence": round(confidence, 4),
        })
```
